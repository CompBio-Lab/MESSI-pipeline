```python
import pandas as pd
```

```python
# Result dir
py_result = "../results/cross_validation/cv_python/merge_result_table/Python-result.csv"
r_result = "../results/cross_validation/cv_r/merge_result_table/R-result.csv"
```

```python
py_df = pd.read_csv(py_result)
r_df = pd.read_csv(r_result)
```

```python
# Merge both dfs
big_df = pd.concat([r_df, py_df])
```

```python
def getY(x, threshold):
    if x >= threshold:
        return 1
    return 0

threshold = 0.5
```

```python
big_df["yhat"] = big_df["phat"].apply(lambda x: getY(x=x, threshold=threshold))
```
